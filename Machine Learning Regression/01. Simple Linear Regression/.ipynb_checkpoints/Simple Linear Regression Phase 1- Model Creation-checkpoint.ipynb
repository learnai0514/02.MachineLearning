```python
# Import pandas library which is used read database 
# Import matlotlibrary which is used to plot graph from dataset
import pandas as pd
import matplotlib.pyplot as plt
```

# Data Preparation

```python
# Read csv data from Salary data set and assign to data set variable
dataset = pd.read_csv("Salary_Data.csv")
```

```python
# display dataset
dataset
```

# Split X and Y

```python
# read/split X coordinate varible into independent variable
independent = dataset[["YearsExperience"]]
```

```python
# read/split Y Coordinate varaibale into dependent variable
dependent = dataset[["Salary"]]
```

```python
# display dependent variable
dependent
```

# Train and Test Split

```python
# plot libray to display chart for the data . in that we can add X Y coorinate label using Plt.Xlable and plt.Ylabel
# plt. show with display the values in the chart
plt.scatter(independent,dependent)
plt.xlabel('YearsExperience', fontsize=20)
plt.ylabel('Salary', fontsize=20)
plt.show()
```

```python
#sklearn library is used to split the date into two set of values. one of training set and another one for test set
# it split the values into 4 sections like test x,y and train x,y  test size is number of data it will taken for validate the trained model
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(independent, dependent, test_size =0.30, random_state=0)
```

# Learning 

```python
#sklearn imports LinearRegression Alogorithm class
from sklearn.linear_model import LinearRegression
# Since it is a class initialize the object
regressor = LinearRegression()
# Fit the train data
regressor.fit(x_train, y_train)
# for linear equation we need to indentify wait and bias coef returns weight 
weight = regressor.coef_
weight
#intercept_ will indetify bias which is origin of the data
bias = regressor.intercept_
bias
# y_pred is prdiction data this is the model output
y_pred = regressor.predict(x_test)
```

```python
# in this library import the evaluation of the model. r2 is the model we are using to validat the model
from sklearn.metrics import r2_score
# if r2 score is closer to 1 this model is sucess and if it 0 it is failure
r_score= r2_score(y_test, y_pred)
# display r score
r_score
```

# Save Model

```python
# Import pickle libray to save the model
import pickle
```

```python
file_name="finalized_model_linear.sav"
```

```python
# this kine it will save the model mode is wb
pickle.dump(regressor,open(file_name, 'wb'))
```

# Load

```python
# reload saved model for deployment since it is read mode is rb
loaded_model = pickle.load(open("finalized_model_linear.sav",'rb'))
#predict is used to test the model
result = loaded_model.predict([[15]])
```
