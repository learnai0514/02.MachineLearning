```python
import pandas as pd
```

```python
dataset = pd.read_csv("50_Startups.csv")
```

```python
dataset
```

```python
dataset = pd.get_dummies(dataset, dtype=int, drop_first=True)
```

```python
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
independent
```

```python
dependent = dataset[['Profit']]
dependent
```

```python
from sklearn.model_selection import train_test_split
x_test, x_train, y_test, y_train = train_test_split(independent, dependent, test_size=0.30, random_state=0)
x_test
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(criterion="poisson", max_features="log2", splitter="random")
regressor = regressor.fit(x_train, y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree
tree.plot_tree(regressor)
plt.show()
```

```python
y_pred = regressor.predict(x_test)
from sklearn.metrics import r2_score
r_score = r2_score(y_test, y_pred)
r_score
```

# Save Model

```python
import pickle
```

```python
file_name = "Finalized_model_DecisionTreeRegression.sav"
pickle.dump(regressor, open(file_name, 'wb'))
```

# Load Model

```python
loaded_model = pickle.load(open(file_name,'rb'))
result = loaded_model.predict([[162597.70,151377.59,443898.53,0,0]])
result
```
